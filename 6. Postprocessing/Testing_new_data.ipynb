```python
import pandas as pd
import geopandas as gpd
import fiona
```

# Testing water bodies

```python
path = r"C:\Users\Carlos\Desktop\MSR\Model-Supply-Regions-MSR-Toolset\1. MSR Creator\Input\Mapas\water_bodies.shp"
water_bodies =gpd.read_file(path)
```

```python
print(water_bodies.crs)
print(water_bodies.total_bounds)
with fiona.open(path) as src:
    print(src.crs)
```

```python
# Input / output paths
input_path = r"C:\Users\Carlos\Desktop\MSR\Model-Supply-Regions-MSR-Toolset\1. MSR Creator\Input\Mapas\water_bodies.shp"
output_path = r"C:\Users\Carlos\Desktop\MSR\Model-Supply-Regions-MSR-Toolset\1. MSR Creator\Input\Mapas\water_bodies_fixed.shp"

# Load dataset
gdf = gpd.read_file(input_path)

# Check CRS
print("Original CRS:", gdf.crs)

# Assign CRS ONLY if missing
if gdf.crs is None:
    print("CRS missing → assigning EPSG:4326")
    gdf = gdf.set_crs("EPSG:4326")
else:
    print("CRS already exists → no change")

# Save corrected file
gdf.to_file(output_path)

print("Saved fixed file to:", output_path)
```
